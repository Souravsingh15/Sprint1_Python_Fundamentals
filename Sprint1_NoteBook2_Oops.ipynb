# Class



A class is a blueprint or template used to create objects. It defines the attributes (data) and methods (functions) that the objects created from it will have. A class itself does not store individual data; instead, it describes what an object should look like and what actions it can perform.

```python
class Student:
    pass

print(Student)
```

### Explanation

In this example, we created a class named **Student**.

The keyword **pass** is used because the class currently has no attributes or methods.

The output shows that a class named **Student** has been created successfully.

# Object



An object is an instance of a class. It is created from a class and contains its own data. Multiple objects can be created from the same class, and each object can store different values while sharing the same structure and behavior defined by the class.

```python
class Student:
    pass

student1 = Student()
student2 = Student()

print(student1)
print(student2)
```

### Explanation

Here, two different objects are created from the same Student class.

Although both belong to the same class, they are different objects and occupy different memory locations.

# Constructor (__init__)



A constructor is a special method that automatically executes whenever an object is created.

In Python, the constructor is written as **__init__()**.

It is mainly used to initialize the object's attributes with initial values.



### Real-World Example

When a new employee joins a company, their name, employee ID and department are recorded immediately.

Similarly, the constructor automatically stores the initial information when an object is created.

```python
class Student:

    def __init__(self, name, age):
        self.name = name
        self.age = age

student1 = Student("Rahul", 22)

print(student1.name)
print(student1.age)
```

### Explanation

The constructor automatically runs when the Student object is created.

The values "Rahul" and 22 are stored inside the object using **self.name** and **self.age**.

# Instance Variables



Instance variables are variables that belong to a specific object.

Each object has its own copy of these variables, so changing one object's data does not affect another object.

```python
class Student:

    def __init__(self, name, age):
        self.name = name
        self.age = age

student1 = Student("Rahul", 22)
student2 = Student("Priya", 21)

print(student1.name)
print(student2.name)
```

### Explanation

Both objects have separate copies of the variables **name** and **age**.

Changing one object's values will not affect the other object.

# Class Variables


Class variables belong to the class itself rather than individual objects.

All objects created from the class share the same class variable.

```python
class Student:

    college = "ABC University"

    def __init__(self, name):
        self.name = name

student1 = Student("Rahul")
student2 = Student("Priya")

print(student1.college)
print(student2.college)
```

### Explanation

The variable **college** belongs to the class.

Every object can access it, and all objects share the same value.

# Instance Methods



Instance methods are functions defined inside a class that operate on individual objects.

They use the **self** parameter to access the object's attributes and perform operations specific to that object.

```python
class Student:

    def __init__(self, name, age):
        self.name = name
        self.age = age

    def introduce(self):
        print(f"My name is {self.name} and I am {self.age} years old.")

student1 = Student("Rahul", 22)

student1.introduce()
```

### Explanation

The method **introduce()** uses the **self** parameter to access the object's own data.

When called, it prints the information stored inside that specific object.

# Static Methods



A static method is a method that belongs to a class but does not access instance variables or class variables. It performs a general task related to the class. Static methods are created using the `@staticmethod` decorator and do not require the `self` or `cls` parameter.

```python
class Student:

    @staticmethod
    def passing_marks():
        print("Passing marks are 40.")

Student.passing_marks()
```

### Explanation

The method `passing_marks()` does not use `self` or `cls`.

It performs a general task related to the class, so it is defined as a static method.

# Class Methods



A class method works with class variables instead of instance variables. It uses the `cls` parameter to refer to the class and is created using the `@classmethod` decorator.

```python
class Student:

    college = "ABC University"

    @classmethod
    def change_college(cls, new_name):
        cls.college = new_name

Student.change_college("XYZ University")

print(Student.college)
```

### Explanation

The method `change_college()` updates the class variable using the `cls` parameter.

All objects automatically use the updated value.

# Properties (@property)


The `@property` decorator allows a method to be accessed like an attribute. It is commonly used to control access to private data while providing a clean interface.

```python
class Student:

    def __init__(self, marks):
        self._marks = marks

    @property
    def percentage(self):
        return (self._marks / 500) * 100

student = Student(425)

print(student.percentage)
```

### Explanation

The `percentage()` method is accessed like an attribute instead of calling it with parentheses.

The calculation happens automatically whenever it is accessed.

# Encapsulation

Encapsulation is the process of combining data and methods into a single unit (class) while restricting direct access to sensitive data. Python uses naming conventions like double underscores (`__`) to indicate private attributes.

```python
class BankAccount:

    def __init__(self, balance):
        self.__balance = balance

    def show_balance(self):
        print("Balance:", self.__balance)

account = BankAccount(5000)

account.show_balance()
```

### Explanation

The variable `__balance` is private.

It cannot be accessed directly from outside the class. Instead, the `show_balance()` method provides controlled access.

# Abstraction



Abstraction means hiding the implementation details and showing only the essential functionality to the user. Python provides abstraction through the `abc` module using abstract classes and abstract methods.



### Real-World Example

When driving a car, the driver only uses the steering wheel, brake and accelerator.

The driver does not need to know how the engine works internally.

This is abstraction.

```python
from abc import ABC, abstractmethod

class Animal(ABC):

    @abstractmethod
    def sound(self):
        pass

class Dog(Animal):

    def sound(self):
        print("Dog barks")

dog = Dog()

dog.sound()
```

### Explanation

`Animal` is an abstract class.

The abstract method `sound()` must be implemented by every child class.

Here, the `Dog` class provides its own implementation of the `sound()` method.

# Inheritance



Inheritance is an Object-Oriented Programming concept where one class acquires the properties and methods of another class. The existing class is called the **Parent (Base) Class**, and the new class is called the **Child (Derived) Class**.

Inheritance promotes code reusability because the child class can use the features of the parent class without rewriting the code.

```python
class Person:

    def __init__(self, name):
        self.name = name

    def display(self):
        print("Name:", self.name)


class Student(Person):

    def __init__(self, name, roll):
        super().__init__(name)
        self.roll = roll

    def show(self):
        print("Roll Number:", self.roll)


student = Student("Rahul", 101)

student.display()
student.show()
```

### Explanation

The Student class inherits from the Person class.

The Student object can access both the parent's method `display()` and its own method `show()`. This avoids rewriting common code.

# Multiple Inheritance

Multiple inheritance allows a class to inherit from more than one parent class.

The child class gains access to the methods and attributes of all parent classes.

```python
class Camera:

    def take_photo(self):
        print("Photo Captured")


class MusicPlayer:

    def play_music(self):
        print("Playing Music")


class Smartphone(Camera, MusicPlayer):
    pass


phone = Smartphone()

phone.take_photo()
phone.play_music()
```

### Explanation

The Smartphone class inherits from both Camera and MusicPlayer.

Therefore, it can use the methods of both parent classes.

# Multilevel Inheritance


Multilevel inheritance occurs when a child class becomes the parent of another class.

This forms a chain of inheritance where properties are inherited through multiple levels.

```python
class Grandfather:

    def grandfather_property(self):
        print("Grandfather's Property")


class Father(Grandfather):

    def father_property(self):
        print("Father's Property")


class Son(Father):

    def son_property(self):
        print("Son's Property")


son = Son()

son.grandfather_property()
son.father_property()
son.son_property()
```

### Explanation

The Son class inherits from Father, and Father inherits from Grandfather.

Therefore, Son can access methods from all three levels.

# Hierarchical Inheritance



Hierarchical inheritance occurs when multiple child classes inherit from the same parent class.

Each child gets access to the common features of the parent while also having its own unique features.

```python
class Person:

    def introduce(self):
        print("I am a person.")


class Student(Person):

    def study(self):
        print("Student is studying.")


class Teacher(Person):

    def teach(self):
        print("Teacher is teaching.")


student = Student()
teacher = Teacher()

student.introduce()
student.study()

teacher.introduce()
teacher.teach()
```

### Explanation

Both Student and Teacher inherit from Person.

Each child class can use the common method `introduce()` while also having its own specific method.

# Polymorphism



Polymorphism means "many forms". It allows the same method name to perform different actions depending on the object that calls it. This makes the code more flexible, reusable, and easier to maintain.

In Python, polymorphism is commonly achieved through method overriding.



### Real-World Example

Different animals make different sounds.

- Dog → Barks
- Cat → Meows
- Cow → Moos

Although the action is the same (making a sound), each animal performs it differently. This is an example of polymorphism.

```python
class Dog:

    def sound(self):
        print("Dog barks")


class Cat:

    def sound(self):
        print("Cat meows")


class Cow:

    def sound(self):
        print("Cow moos")


animals = [Dog(), Cat(), Cow()]

for animal in animals:
    animal.sound()
```

### Explanation

Each class defines its own version of the `sound()` method.

During iteration, the same method name is called for every object, but the output depends on the object's class.

This demonstrates polymorphism.

# Method Overriding

Method overriding occurs when a child class provides its own implementation of a method that already exists in the parent class.

The child class replaces the parent's implementation while keeping the same method name.

```python
class Animal:

    def sound(self):
        print("Animals make sounds")


class Dog(Animal):

    def sound(self):
        print("Dog barks")


dog = Dog()

dog.sound()
```

# Method Overloading (Python Approach)


Method overloading means having multiple methods with the same name but different parameters.

Unlike Java or C++, Python does not support method overloading directly.

Instead, Python achieves similar behavior using default arguments, variable-length arguments (`*args`), or conditional logic inside a single method.



### Real-World Example

Suppose a calculator performs addition.

It should be able to add:

- Two numbers
- Three numbers
- Four numbers

Instead of creating multiple methods, one method handles different numbers of inputs.

```python
class Calculator:

    def add(self, *numbers):
        return sum(numbers)


calc = Calculator()

print(calc.add(10, 20))
print(calc.add(10, 20, 30))
print(calc.add(10, 20, 30, 40))
```

### Explanation

Python does not support method overloading directly.

Here, the `*numbers` parameter allows the method to accept any number of arguments.

This provides functionality similar to method overloading.

# Composition


Composition is a relationship where one class completely owns another class. The contained object cannot exist independently of the container object.

If the container object is destroyed, the contained object is also destroyed. Composition represents a strong "has-a" relationship.



### Real-World Example

A house has rooms.

A room usually belongs to a specific house. If the house no longer exists, the rooms are also considered part of that house.

This is an example of composition.

```python
class Engine:

    def start(self):
        print("Engine Started")


class Car:

    def __init__(self):
        self.engine = Engine()

    def drive(self):
        self.engine.start()
        print("Car is moving")


car = Car()
car.drive()
```

# Aggregation



Aggregation is a relationship where one class uses another class, but both classes can exist independently.

The contained object is created outside the container class and is simply passed to it.

Aggregation represents a weak "has-a" relationship.

```python
class Teacher:

    def __init__(self, name):
        self.name = name


class Department:

    def __init__(self, teacher):
        self.teacher = teacher

    def display(self):
        print("Teacher:", self.teacher.name)


teacher = Teacher("Rahul")

department = Department(teacher)

department.display()
```

### Explanation

The Teacher object is created separately and then passed to the Department.

Since both objects can exist independently, this relationship is called aggregation.

# Association



Association is a relationship where two independent classes interact with each other.

Neither class owns the other, and both can exist independently.

Association represents a general relationship between objects.

```python
class Customer:

    def __init__(self, name):
        self.name = name


class Order:

    def __init__(self, order_id):
        self.order_id = order_id

    def place_order(self, customer):
        print(f"Order {self.order_id} placed by {customer.name}")


customer = Customer("Rahul")
order = Order(101)

order.place_order(customer)
```

### Explanation

The Customer and Order classes are independent.

The Order class simply interacts with a Customer object through the `place_order()` method.

This demonstrates association.

# super()



The `super()` function is used to access the methods and constructor of the parent class from the child class.

It helps avoid rewriting code and supports inheritance.

```python
class Person:

    def __init__(self, name):
        self.name = name


class Student(Person):

    def __init__(self, name, roll):
        super().__init__(name)
        self.roll = roll


student = Student("Rahul",101)

print(student.name)
print(student.roll)
```

### Explanation

The `super()` function calls the constructor of the Person class.

This avoids duplicate code and initializes the inherited attributes.

# __str__()



The `__str__()` method defines how an object should be displayed when passed to the `print()` function.

It provides a user-friendly string representation of the object.


### Real-World Example

Instead of displaying a person's ID number, it is more useful to display their name.

Similarly, `__str__()` controls what users see when printing an object.

```python
class Student:

    def __init__(self,name):
        self.name=name

    def __str__(self):
        return f"Student Name: {self.name}"

student=Student("Rahul")

print(student)
```

### Explanation

The `__str__()` method returns a readable description of the object.

When `print(student)` is executed, Python automatically calls `__str__()`.

# __repr__()



The `__repr__()` method returns the official string representation of an object.

It is mainly intended for developers and debugging.



### Real-World Example

Suppose a mechanic needs the complete engine details instead of a simple car name.

Similarly, `__repr__()` provides more detailed information for developers.

```python
class Student:

    def __init__(self,name):
        self.name=name

    def __repr__(self):
        return f"Student('{self.name}')"

student=Student("Rahul")

print(student)
```

### Explanation

The `__repr__()` method returns a detailed representation of the object.

It is mainly used for debugging and development purposes.

# __len__()



The `__len__()` method allows an object to define its own length.

When the built-in `len()` function is called, Python automatically executes the `__len__()` method.

```python
class Library:

    def __init__(self, books):
        self.books = books

    def __len__(self):
        return len(self.books)


library = Library(["Python", "Java", "SQL"])

print(len(library))
```

### Explanation

The `__len__()` method tells Python how to calculate the length of the Library object.

Calling `len(library)` automatically invokes `__len__()`.

# Dataclasses (Bonus)


A dataclass is a special feature introduced in Python that automatically generates common methods such as `__init__()`, `__repr__()`, and `__eq__()`.

It helps reduce boilerplate code when a class is mainly used to store data.



### Real-World Example

Suppose a school stores information about students such as name, roll number, and marks.

Instead of writing the constructor and other methods manually every time, a dataclass automatically creates them.

This makes the code shorter, cleaner, and easier to maintain.

```python
from dataclasses import dataclass

@dataclass
class Student:
    name: str
    age: int
    marks: int

student = Student("Rahul", 21, 90)

print(student)
```

### Explanation

The `@dataclass` decorator automatically creates the constructor and a readable object representation.

We only declare the attributes, and Python generates the required methods for us.

This reduces the amount of code we need to write.

# Singleton Pattern (Bonus)



The Singleton Pattern is a design pattern that ensures only one object of a class can be created.

If another object is requested, the same existing object is returned instead of creating a new one.

This pattern is useful when exactly one shared object is required throughout the application.

```python
class Singleton:

    _instance = None

    def __new__(cls):

        if cls._instance is None:
            cls._instance = super().__new__(cls)

        return cls._instance


obj1 = Singleton()
obj2 = Singleton()

print(obj1 is obj2)
```

### Explanation

The `__new__()` method controls object creation.

When the first object is created, it is stored in `_instance`.

If another object is requested, the existing object is returned instead of creating a new one.

The expression `obj1 is obj2` returns `True`, proving that both variables refer to the same object.
