# What is Logging?



Logging is the process of recording important events, messages, warnings, and errors that occur while a program is running.

Instead of displaying messages only on the screen, logging stores them in a structured way so they can be reviewed later for debugging, monitoring, and maintenance.


### Real-World Example

Imagine a security guard maintaining a visitor register.

Every visitor's entry and exit time is recorded.

Later, if an issue occurs, the register helps identify what happened.

Similarly, logging keeps a record of important events in a program.

```python
import logging

logging.basicConfig(level=logging.INFO)

logging.info("Program started.")
logging.info("User logged in.")
```

### Explanation

The `logging` module is imported to record program events.

The `INFO` level is set, and two informational messages are logged.

# Why use Logging?



Logging helps developers understand what happens inside a program while it is running.

It is useful for debugging errors, monitoring applications, tracking events, and maintaining software after deployment.



### Real-World Example

A CCTV camera continuously records activities inside a building.

If an incident occurs, the recorded footage helps identify the cause.

Similarly, logging records events that help developers analyze program behavior.

```python
import logging

logging.basicConfig(level=logging.INFO)

logging.info("Application started.")
logging.warning("Low disk space.")
logging.error("Database connection failed.")
```

### Explanation

Logging records different types of events, allowing developers to identify issues without manually checking the program.

# print() vs logging



Both `print()` and `logging` display information, but they serve different purposes.

`print()` is mainly used for temporary output during learning or simple debugging.

`logging` is designed for real applications because it supports log levels, timestamps, file storage, and better control over messages.

```python
import logging

logging.basicConfig(level=logging.INFO)

print("Program Started")

logging.info("Program Started")
```

### Explanation

`print()` only displays the message on the screen.

`logging.info()` records the message as an official log entry.

# Log Levels



Log levels indicate the importance or severity of log messages.

They help developers filter and organize logs based on their priority.



### Real-World Example

In a hospital:

- Normal patient update → INFO
- Minor issue → WARNING
- Serious condition → ERROR
- Life-threatening emergency → CRITICAL

Similarly, logging uses different levels to classify messages.

```python
import logging

logging.basicConfig(level=logging.DEBUG)

logging.debug("Debug Message")
logging.info("Information Message")
logging.warning("Warning Message")
logging.error("Error Message")
logging.critical("Critical Message")
```

# DEBUG



The `DEBUG` log level is used to record detailed information about the program's execution.

It is mainly used during development to understand how the program is working step by step and to identify the cause of bugs.

```python
import logging

logging.basicConfig(level=logging.DEBUG)

logging.debug("Variable x = 10")
logging.debug("Program entered the login function.")
```

### Explanation

The `DEBUG` level records detailed information about the program.

These messages are mainly useful while developing and debugging applications.

# INFO



The `INFO` log level is used to record normal events that confirm the program is working as expected.

These messages provide useful information without indicating any problem.

```python
import logging

logging.basicConfig(level=logging.INFO)

logging.info("Application started successfully.")
logging.info("User logged in.")
```

### Explanation

The `INFO` level records normal events that help monitor the application's execution.

# WARNING



The `WARNING` log level indicates that something unexpected has happened, but the program can still continue running.

Warnings help developers identify potential problems before they become serious.



### Real-World Example

Suppose your phone battery reaches 15%.

The phone displays a warning, but it continues working.

Similarly, the `WARNING` level reports situations that require attention but are not critical.

```python
import logging

logging.basicConfig(level=logging.WARNING)

logging.warning("Battery level is low.")
```

### Explanation

The `WARNING` level records messages that indicate possible issues while allowing the application to continue running.

# ERROR



The `ERROR` log level indicates that a significant problem has occurred.

The application may not be able to complete a particular task, although it may continue running.

```python
import logging

logging.basicConfig(level=logging.ERROR)

logging.error("Database connection failed.")
```

# CRITICAL



The `CRITICAL` log level represents the most severe errors.

It indicates that the application has encountered a serious problem and may not be able to continue running.



### Real-World Example

Suppose the server hosting an online shopping website crashes.

Customers cannot place orders until the server is restored.

This is a critical situation.

```python
import logging

logging.basicConfig(level=logging.CRITICAL)

logging.critical("Server has crashed.")
```

### Explanation

The `CRITICAL` level records severe problems that require immediate attention because the application may stop functioning properly.

## Log Levels Summary

| Log Level | Purpose | Example |
|------------|----------|---------|
| DEBUG | Detailed information for debugging | Variable values |
| INFO | Normal application events | User logged in |
| WARNING | Potential issues | Low disk space |
| ERROR | Significant problems | Database connection failed |
| CRITICAL | Serious failures | Server crashed |

# Log Formatters



A log formatter defines how log messages should appear.

Instead of displaying only the message, a formatter can include additional information such as the date, time, log level, and message.

This makes log files easier to read and analyze.


### Real-World Example

Imagine a newspaper.

Every news article includes the date, time, headline, and details.

Similarly, a log formatter organizes log messages into a consistent format.

```python
import logging

logging.basicConfig(
    level=logging.INFO,
    format="%(asctime)s - %(levelname)s - %(message)s"
)

logging.info("Application Started")
logging.warning("Low Memory")
```

# File Handlers



A File Handler writes log messages into a file instead of displaying them only on the console.

This allows developers to review logs later, even after the program has finished running.



### Real-World Example

A company keeps employee attendance records in a register.

The information remains available even after employees leave.

Similarly, a File Handler stores log messages permanently in a log file.

```python
import logging

logging.basicConfig(
    filename="application.log",
    level=logging.INFO
)

logging.info("Program Started")
logging.error("An error occurred")
```

# Console Handlers



A Console Handler displays log messages directly on the terminal or console while the program is running.

It is useful during development because developers can immediately see program events.

```python
import logging

logger = logging.getLogger()

console = logging.StreamHandler()

logger.addHandler(console)

logger.setLevel(logging.INFO)

logger.info("Application Started")
```

# Rotating Logs (Bonus)



Rotating logs automatically create new log files when the current log file becomes too large.

This prevents log files from growing indefinitely and helps manage storage efficiently.

```python
import logging
from logging.handlers import RotatingFileHandler

handler = RotatingFileHandler(
    "application.log",
    maxBytes=500,
    backupCount=2
)

logger = logging.getLogger()

logger.addHandler(handler)

logger.setLevel(logging.INFO)

logger.info("Logging using Rotating File Handler")
```

### Explanation

`RotatingFileHandler` creates a new log file when the current file exceeds the specified size.

`backupCount` determines how many old log files should be retained.

# Logging Best Practices



Logging should provide useful information without exposing sensitive data.

Well-written logs make debugging, monitoring, and maintaining applications much easier.



### Real-World Example

A hospital records patient information carefully.

Only necessary details are written, while confidential information remains protected.

Similarly, application logs should contain useful information without exposing sensitive data.

```python
import logging

logging.basicConfig(level=logging.INFO)

user = "Rahul"

logging.info(f"User {user} logged in successfully.")
```

## Best Practices

- Use appropriate log levels (DEBUG, INFO, WARNING, ERROR, CRITICAL).
- Avoid using `print()` for production applications.
- Do not log sensitive information such as passwords or API keys.
- Use meaningful log messages.
- Store logs in files for future analysis.
- Use log rotation to prevent very large log files.
- Include timestamps in log messages.
