# try



The `try` block contains the code that might produce an exception during execution.

Python executes the code inside the `try` block first. If no exception occurs, the program continues normally. If an exception occurs, Python stops executing the remaining statements in the `try` block and looks for a matching `except` block.


### Real-World Example

Imagine you are driving a car on a road.

Normally, you continue driving without any issues.

However, there is always a possibility of encountering an obstacle or a flat tire.

Similarly, the `try` block contains code that is expected to run normally but may sometimes produce an exception.

```python
try:
    num = int(input("Enter a number: "))
    print("You entered:", num)
except:
    print("An error occurred.")
```

### Explanation

The program attempts to convert the user's input into an integer.

If the input is a valid number, it prints the number.

If the input cannot be converted into an integer, an exception occurs, and Python transfers control to the `except` block.

# except


The `except` block handles exceptions that occur inside the `try` block.

Instead of terminating the program abruptly, it allows us to display a meaningful message or take an alternative action.



### Real-World Example

Suppose you withdraw money from an ATM.

If the machine runs out of cash, instead of stopping unexpectedly, it displays a message such as "Transaction cannot be completed."

Similarly, the `except` block handles unexpected situations gracefully.

# else



The `else` block executes only if no exception occurs in the `try` block.

It is useful for writing code that should run only after the `try` block has completed successfully.

# finally



The `finally` block always executes, whether an exception occurs or not.

It is commonly used for cleanup tasks such as closing files, releasing resources, or disconnecting from a database.

### Explanation

The `finally` block executes regardless of whether an exception occurs.

It is mainly used to perform cleanup operations before the program continues or exits.

# raise



The `raise` keyword is used to manually generate an exception in Python.

Normally, Python raises exceptions automatically when an error occurs. However, there are situations where we want to create our own exception based on certain conditions.

Using `raise` allows us to stop the program and display a meaningful error message when a condition is not satisfied.

age = int(input("Enter your age: "))

if age < 18:
    raise ValueError("Age must be 18 or above.")

print("You are eligible to vote.")

### Explanation

The program checks whether the entered age is less than 18.

If the condition is true, the `raise` statement manually generates a `ValueError` with a custom message.

If the condition is false, the program continues normally.

# assert



The `assert` statement is used to check whether a condition is true.

If the condition is true, the program continues normally.

If the condition is false, Python raises an `AssertionError`.

Assertions are commonly used during development to detect logical errors and verify assumptions.

### Explanation

The `assert` statement checks whether `marks >= 40`.

If the condition is true, the program prints the success message.

If the condition is false, Python raises an `AssertionError` with the specified message.

# Custom Exceptions



Python allows us to create our own exception classes, known as custom exceptions.

A custom exception is created by defining a new class that inherits from the built-in `Exception` class.

Custom exceptions help make programs easier to understand by providing meaningful error messages for application-specific situations.

class WithdrawalLimitExceeded(Exception):
    pass


amount = int(input("Enter withdrawal amount: "))

if amount > 20000:
    raise WithdrawalLimitExceeded("Withdrawal limit exceeded.")

print("Transaction Successful.")




### Explanation

The class `WithdrawalLimitExceeded` is a custom exception because it inherits from the built-in `Exception` class.

When the withdrawal amount exceeds ₹20,000, the program raises the custom exception with a meaningful message.

Otherwise, the transaction completes successfully.

## Difference Between raise, assert and Custom Exceptions

| Feature | raise | assert | Custom Exception |
|---------|--------|---------|------------------|
| Purpose | Manually raise an exception | Verify a condition | Create your own exception type |
| Used For | Invalid conditions | Debugging and testing | Application-specific errors |
| Raises | Any exception | AssertionError | User-defined exception |
| Example | Age validation | Marks validation | ATM withdrawal limit |

# ValueError



A `ValueError` occurs when a function receives an argument of the correct data type but with an invalid value.



### Real-World Example

Suppose a form asks you to enter your age as a number.

If you enter "abc", the value is invalid even though the program expects a number.

This results in a ValueError.

### Explanation

If the user enters text instead of a number, Python raises a ValueError because the value cannot be converted into an integer.

# TypeError



A `TypeError` occurs when an operation is performed on incompatible data types.

except TypeError:

### Explanation

Python cannot add an integer and a string together, so it raises a TypeError.
