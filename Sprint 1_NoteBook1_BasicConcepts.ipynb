# if Statement

## Step 1 – Understand the Concept

The `if` statement is used to make decisions in a Python program. It checks whether a condition is True or False. If the condition is True, the code inside the `if` block is executed. If the condition is False, Python skips the block and continues with the rest of the program.

The `if` statement is useful when we want a program to perform an action only when a specific condition is satisfied.

```python
# Step 3 – Implement the Concept

age = 20

if age >= 18:
    print("You are eligible to vote.")
```

# if...else Statement

## Step 1 – Understand the Concept

The `if...else` statement is used when there are two possible outcomes. If the condition is True, the `if` block executes. Otherwise, the `else` block executes.

It helps programs choose between two different actions.

```python
# Step 3 – Implement the Concept

marks = 40

if marks >= 35:
    print("Result: Pass")
else:
    print("Result: Fail")
```

# if...elif...else Statement


The `if...elif...else` statement is used when there are multiple conditions to check. Python evaluates each condition one by one. As soon as it finds a True condition, it executes the corresponding block and skips the remaining conditions.

This helps in making decisions when there are more than two possible outcomes.

```python
# Step 3 – Implement the Concept

marks = 82

if marks >= 90:
    print("Grade A")
elif marks >= 75:
    print("Grade B")
elif marks >= 50:
    print("Grade C")
else:
    print("Fail")
```

# for Loop

## Step 1 – Understand the Concept

A `for` loop is used to repeat a block of code for each item in a sequence such as a list, tuple, string, or range of numbers. It is useful when we know how many times we want the loop to execute or when we want to iterate through a collection.

Using a `for` loop reduces repetitive code and makes programs shorter and easier to understand.

```python
# Implement the Concept

students = ["Rahul", "Priya", "Sourav", "Aman"]

for student in students:
    print("Present:", student)
```

### Explanation

The list `students` contains four names.

The `for` loop visits each name one by one.

During every iteration, the current student's name is stored in the variable `student` and printed.

This demonstrates how a `for` loop automatically repeats the same code for every item in a sequence.

# while Loop



A `while` loop repeatedly executes a block of code as long as a specified condition remains True. Unlike a `for` loop, a `while` loop is useful when we do not know in advance how many times the loop should run.

The condition is checked before every iteration. When it becomes False, the loop stops.

```python
#  Implement the Concept

count = 1

while count <= 5:
    print("Count:", count)
    count += 1
```

### Explanation

The variable `count` starts from 1.

The loop continues while `count` is less than or equal to 5.

After each iteration, the value of `count` increases by 1.

When `count` becomes 6, the condition becomes False, and the loop stops.

This demonstrates how a `while` loop executes based on a condition.

# break Statement



The `break` statement is used to immediately stop a loop before it finishes all its iterations. Once `break` is executed, the program exits the loop and continues with the next statement after the loop.

It is useful when the required result has already been found and further looping is unnecessary.

```python
# Step 3 – Implement the Concept

for number in range(1, 11):

    if number == 6:
        break

    print(number)
```

### Explanation

The loop starts printing numbers from 1.

When the value becomes 6, the `break` statement is executed.

The loop immediately stops, so only the numbers 1 to 5 are printed.

This demonstrates how `break` terminates a loop before it completes all iterations.

# continue Statement



The `continue` statement is used to skip the current iteration of a loop and move directly to the next iteration. Unlike `break`, it does not stop the entire loop.

It is useful when certain values should be ignored while processing the remaining data.

```python
# Step 3 – Implement the Concept

for number in range(1, 6):

    if number == 3:
        continue

    print(number)
```

# pass Statement



The `pass` statement is a placeholder that performs no action. It is used when Python requires a statement syntactically, but we do not want to write the actual implementation yet.

It allows us to create incomplete code without causing syntax errors.

```python
# Step 3 – Implement the Concept

for number in range(1, 6):

    if number == 3:
        pass

    print(number)
```

The program continues normally and prints all numbers from 1 to 5.

This demonstrates that `pass` is simply a placeholder and does not affect program execution.

# List



A list is an ordered and mutable collection in Python. It can store multiple values of different data types in a single variable. Since lists are mutable, we can add, remove, or modify elements after the list has been created.

Lists are commonly used when the data needs to change during program execution.

```python


fruits = ["Apple", "Banana", "Orange"]

print("Original List:", fruits)

fruits.append("Mango")

print("Updated List:", fruits)

fruits[1] = "Grapes"

print("Modified List:", fruits)
```

### Explanation

The program creates a list containing three fruits.

The `append()` method adds a new fruit to the list.

The second element is then replaced with another value.

This demonstrates that lists are mutable because their contents can be modified after creation.

# Tuple



A tuple is an ordered collection of values that cannot be modified after it is created. This means we cannot add, remove, or change its elements.

Tuples are useful when the stored data should remain constant throughout the program.

```python


student = ("Sourav", 22, "B.Tech")

print(student)

print("Name:", student[0])

print("Age:", student[1])
```

# Keyword Arguments


Keyword arguments allow us to pass values to a function by specifying the parameter names during the function call. This makes the code more readable and allows the arguments to be passed in any order, as long as the correct parameter names are used.

# Variable-Length Arguments (*args and **kwargs)



Sometimes we do not know how many arguments will be passed to a function. Python provides `*args` to accept multiple positional arguments and `**kwargs` to accept multiple keyword arguments.

`*args` stores extra positional arguments as a tuple, while `**kwargs` stores extra keyword arguments as a dictionary.

These features make functions more flexible because they can accept any number of inputs.

```python

def calculate_sum(*numbers):
    total = 0

    for number in numbers:
        total += number

    print("Sum =", total)

calculate_sum(10, 20)

calculate_sum(5, 10, 15, 20)
```

### Explanation

The function uses `*numbers` to accept any number of positional arguments.

All the values are stored in a tuple.

The `for` loop adds all the numbers together and prints the total.

This demonstrates how `*args` makes a function flexible enough to accept different numbers of inputs.

```python
# Example of **kwargs

def student_info(**details):

    for key, value in details.items():
        print(key, ":", value)

student_info(Name="Sourav", Age=22, Course="B.Tech")
```

### Explanation

The function uses `**details` to accept any number of keyword arguments.

All the values are stored in a dictionary.

The `items()` method is used to display each key-value pair.

This demonstrates how `**kwargs` accepts flexible keyword arguments.

# Recursive Functions



A recursive function is a function that calls itself to solve a problem. Every recursive function must have a base case that stops the recursion. Without a base case, the function would continue calling itself forever.

Recursion is useful for solving problems that can be divided into smaller versions of the same problem.

```python


def factorial(number):

    if number == 1:
        return 1

    return number * factorial(number - 1)

result = factorial(5)

print("Factorial =", result)
```

### Explanation

The function calculates the factorial of a number.

The base case is when the number becomes 1.

For every other value, the function calls itself with a smaller number.

The recursive calls continue until the base case is reached, after which the results are multiplied together.

This demonstrates how recursion solves a problem by repeatedly breaking it into smaller subproblems.

# Iterators


An iterator is an object that allows us to access the elements of a collection one at a time without directly using indexing. Python automatically uses iterators when we loop through lists, tuples, strings, and other iterable objects.

An iterator remembers its current position and returns the next element whenever the `next()` function is called. Once all elements have been returned, it raises a `StopIteration` exception.

```python

fruits = ["Apple", "Banana", "Mango"]

fruit_iterator = iter(fruits)

print(next(fruit_iterator))
print(next(fruit_iterator))
print(next(fruit_iterator))
```

### Explanation

The `iter()` function converts the list into an iterator.

Each call to `next()` returns the next element from the iterator.

The iterator remembers its position automatically.

This demonstrates how iterators provide one value at a time instead of returning all values together.

# Generators



A generator is a special type of function that produces values one at a time instead of returning all values at once. Unlike a normal function, a generator uses the `yield` keyword instead of `return`.

Generators are memory-efficient because they generate values only when they are needed.

```python


def numbers():
    yield 1
    yield 2
    yield 3

generator = numbers()

print(next(generator))
print(next(generator))
print(next(generator))
```

### Explanation

The function uses the `yield` keyword instead of `return`.

Each call to `next()` produces one value.

The function pauses after every `yield` and resumes from the same point during the next call.

This demonstrates how generators create values one at a time.

# yield Keyword



The `yield` keyword is used inside a generator function to produce one value at a time. Unlike `return`, which ends the function completely, `yield` pauses the function and saves its current state.

When the generator is called again, execution continues from where it previously stopped.

```python

def countdown():

    print("Three")
    yield 3

    print("Two")
    yield 2

    print("One")
    yield 1

counter = countdown()

print(next(counter))
print(next(counter))
print(next(counter))
```

### Explanation

The generator pauses whenever it reaches a `yield` statement.

Each call to `next()` resumes execution from the previous pause.

This demonstrates that `yield` allows a function to produce values gradually while remembering its execution state.

# Decorators


A decorator is a function that modifies or extends the behavior of another function without changing its original code. Decorators are useful for adding extra functionality such as logging, authentication, or timing.

A decorator takes another function as input, adds additional behavior, and returns the modified function.

```python


def decorator_function(function):

    def wrapper():
        print("Program Started")

        function()

        print("Program Finished")

    return wrapper


@decorator_function
def greet():
    print("Hello, Welcome!")

greet()
```

### Explanation

The function `decorator_function` acts as a decorator.

The `wrapper()` function executes additional code before and after the original function.

The `@decorator_function` syntax applies the decorator to `greet()`.

When `greet()` is called, Python first executes the decorator and then the original function.

This demonstrates how decorators extend the functionality of existing functions without modifying their original implementation.

# Closures



A closure is a function defined inside another function that remembers the variables of the outer function even after the outer function has finished executing. This allows the inner function to continue using those variables later.

Closures are useful when we want to preserve data without using global variables.

```python
# Step 3 – Implement the Concept

def outer(message):

    def inner():
        print(message)

    return inner

greeting = outer("Hello, Sourav!")

greeting()
```

### Explanation

The function `outer()` creates another function called `inner()`.

The `inner()` function uses the variable `message` even after `outer()` has finished executing.

This demonstrates how a closure remembers variables from its enclosing scope.

# global Keyword


The `global` keyword allows a function to modify a variable that is defined outside the function. Without using `global`, assigning a value inside a function creates a new local variable instead of changing the global one.

The `global` keyword should be used carefully because changing global variables can make programs harder to understand.

```python

count = 10

def update_count():
    global count
    count = count + 5

update_count()

print("Updated Count:", count)
```

### Explanation

The variable `count` is created outside the function.

Inside the function, the `global` keyword allows the function to modify the same variable.

After calling the function, the value of `count` changes from 10 to 15.

This demonstrates how the `global` keyword works.

# nonlocal Keyword



The `nonlocal` keyword is used inside nested functions to modify a variable that belongs to the enclosing (outer) function. Unlike `global`, it does not affect variables outside the enclosing function.

It is useful when an inner function needs to update a variable from its outer function.

```python
# Step 3 – Implement the Concept

def outer():

    message = "Hello"

    def inner():
        nonlocal message
        message = "Welcome"

    inner()

    print(message)

outer()
```

### Explanation

The variable `message` belongs to the outer function.

The inner function uses `nonlocal` to modify it.

After the inner function executes, the updated value is printed.

This demonstrates how `nonlocal` changes variables from the enclosing function.

# map()



The `map()` function is used to apply the same function to every element of an iterable like a list or tuple. Instead of using a loop to modify each element one by one, `map()` automatically applies the function to every item and returns a map object. We usually convert it into a list to see the results.

```python
marks = [60, 70, 80, 90]

updated_marks = list(map(lambda x: x + 5, marks))

print("Original Marks:", marks)
print("Updated Marks:", updated_marks)
```

### Explanation

This program increases every student's marks by 5 using `map()`. The lambda function performs the addition, and `map()` applies it to every element in the list.

# filter()

```python
numbers = [10, 15, 20, 25, 30, 35]

even_numbers = list(filter(lambda x: x % 2 == 0, numbers))

print("Original List:", numbers)
print("Even Numbers:", even_numbers)
```

### Explanation

The lambda function checks whether a number is even. `filter()` returns only the numbers that satisfy this condition.

# id()
