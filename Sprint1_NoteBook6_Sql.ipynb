## 1. Database

## 2. Tables

A table stores data in rows and columns.

Columns represent the type of information being stored, while rows represent individual records.

### Demonstration

For example, a students table can contain:

| id | name | age |
|---|---|---|
| 1 | Sourav | 25 |
| 2 | Rahul | 24 |

Here, id, name, and age are columns, while each student represents a row.

```python
import sqlite3

connection = sqlite3.connect("college.db")
cursor = connection.cursor()

cursor.execute("""
CREATE TABLE IF NOT EXISTS students (
    id INTEGER,
    name TEXT,
    age INTEGER
)
""")

connection.commit()
connection.close()

print("Students table created.")
```

### Explanation

The program creates a `students` table with three columns: id, name, and age.

The `CREATE TABLE` SQL command is used to create the table.

## 3. Primary Key
A primary key is a column that uniquely identifies each record in a table.

Each value in a primary key must be unique.

### Demonstration

For example, each student can have a unique student ID.

Student 1 → ID 101  
Student 2 → ID 102  
Student 3 → ID 103

The student ID helps us identify each student uniquely.

```python
import sqlite3

connection = sqlite3.connect("college.db")
cursor = connection.cursor()

cursor.execute("""
CREATE TABLE IF NOT EXISTS students_pk (
    student_id INTEGER PRIMARY KEY,
    name TEXT,
    age INTEGER
)
""")

connection.commit()
connection.close()

print("Table with primary key created.")
```

### Explanation

The `student_id` column is defined as the `PRIMARY KEY`.

It uniquely identifies every student record and does not allow duplicate IDs.

## 4. Foreign Key

A foreign key is a column that connects one table to another table.

It usually refers to the primary key of another table.

### Demonstration

Suppose we have a Students table and a Courses table.

The Students table has `student_id` as its primary key.

The Courses table can use `student_id` as a foreign key to connect a course with a student.

```python
import sqlite3

connection = sqlite3.connect("college.db")
cursor = connection.cursor()

cursor.execute("PRAGMA foreign_keys = ON")

cursor.execute("""
CREATE TABLE IF NOT EXISTS courses (
    course_id INTEGER PRIMARY KEY,
    course_name TEXT,
    student_id INTEGER,
    FOREIGN KEY (student_id) REFERENCES students_pk(student_id)
)
""")

connection.commit()
connection.close()

print("Foreign key created.")
```

### Explanation

The `student_id` in the `courses` table is a foreign key.

It references the `student_id` in the `students_pk` table and creates a connection between the two tables.

## 5. Relationships

Relationships describe how tables are connected to each other.

The main types are:

- One-to-One
- One-to-Many
- Many-to-Many

### Demonstration

One-to-One:

One person can have one passport.

One-to-Many:

One teacher can teach many students.

Many-to-Many:

Many students can enroll in many courses.

```python
import sqlite3

connection = sqlite3.connect("college.db")
cursor = connection.cursor()

cursor.execute("""
CREATE TABLE IF NOT EXISTS enrollments (
    student_id INTEGER,
    course_id INTEGER
)
""")

connection.commit()
connection.close()

print("Relationship table created.")
```

### Explanation

The `enrollments` table connects students and courses.

A student can enroll in multiple courses, and a course can have multiple students.

This represents a many-to-many relationship.

## 6. Constraints

Constraints are rules applied to table columns to control what data can be stored.

Common constraints include:

- PRIMARY KEY
- FOREIGN KEY
- NOT NULL
- UNIQUE
- CHECK
- DEFAULT

### Demonstration

For example:

- A student ID should be unique.
- A student's name should not be empty.
- An email should not be repeated.
- Age should be greater than zero.

Constraints help keep database data valid.

```python
import sqlite3

connection = sqlite3.connect("college.db")
cursor = connection.cursor()

cursor.execute("""
CREATE TABLE IF NOT EXISTS students_constraints (
    id INTEGER PRIMARY KEY,
    name TEXT NOT NULL,
    email TEXT UNIQUE,
    age INTEGER CHECK(age > 0),
    city TEXT DEFAULT 'Unknown'
)
""")

connection.commit()
connection.close()

print("Table with constraints created.")
```

### Explanation

The table uses different constraints:

- `PRIMARY KEY` makes the ID unique.
- `NOT NULL` requires a value.
- `UNIQUE` prevents duplicate emails.
- `CHECK` ensures the age is greater than zero.
- `DEFAULT` provides a value when no city is given.

## 7. SELECT

SELECT is used to retrieve data from a table.

We can use SELECT to retrieve specific columns or all columns.

### Demonstration

For example:

SELECT * FROM students;

The `*` means we want to retrieve all columns from the students table.

```python
import sqlite3

connection = sqlite3.connect("college.db")
cursor = connection.cursor()

cursor.execute("SELECT * FROM students")

rows = cursor.fetchall()

for row in rows:
    print(row)

connection.close()
```

### Explanation

The program retrieves all records from the `students` table.

`fetchall()` gets all the records returned by the SELECT query.

## 8. WHERE

WHERE is used to filter records based on a condition.

It allows us to retrieve only the rows that match a specific condition.

### Demonstration

For example:

SELECT * FROM students
WHERE age > 20;

This returns only students whose age is greater than 20.

```python
import sqlite3

connection = sqlite3.connect("college.db")
cursor = connection.cursor()

cursor.execute("SELECT * FROM students WHERE age > 20")

for row in cursor.fetchall():
    print(row)

connection.close()
```

### Explanation

The WHERE clause checks the age of each student.

Only students whose age is greater than 20 are returned.

## 9. ORDER BY

ORDER BY is used to sort the results of a query.

We can sort data in ascending order using ASC or descending order using DESC.

### Demonstration

For example:

SELECT * FROM students
ORDER BY age ASC;

This sorts students from the youngest to the oldest.

```python
import sqlite3

connection = sqlite3.connect("college.db")
cursor = connection.cursor()

cursor.execute("SELECT * FROM students ORDER BY age ASC")

for row in cursor.fetchall():
    print(row)

connection.close()
```

### Explanation

The ORDER BY clause sorts the students according to their age.

ASC sorts from smallest to largest, while DESC sorts from largest to smallest.

## 10. GROUP BY

GROUP BY is used to group rows that have the same value.

It is commonly used with aggregate functions such as COUNT(), SUM(), AVG(), MAX(), and MIN().

### Demonstration

For example:

If several students belong to the same city, GROUP BY can group students according to their city.

We can then count how many students are in each city.

```python
import sqlite3

connection = sqlite3.connect("college.db")
cursor = connection.cursor()

cursor.execute("""
SELECT age, COUNT(*)
FROM students
GROUP BY age
""")

for row in cursor.fetchall():
    print(row)

connection.close()
```

### Explanation

The query groups students according to their city.

COUNT(*) counts the number of students in each city.

## 11. HAVING

HAVING is used to filter grouped results.

It is commonly used together with GROUP BY and aggregate functions.

### Demonstration

For example:

We can group students by age and display only age groups that contain more than one student.

HAVING is used to apply the condition after grouping.

```python
import sqlite3

connection = sqlite3.connect("college.db")
cursor = connection.cursor()

cursor.execute("""
SELECT age, COUNT(*)
FROM students
GROUP BY age
HAVING COUNT(*) > 1
""")

for row in cursor.fetchall():
    print(row)

connection.close()
```

### Explanation

GROUP BY creates groups based on age.

HAVING COUNT(*) > 1 keeps only the age groups that contain more than one student.

SELECT DISTINCT age FROM students;

### Explanation

DISTINCT removes duplicate age values from the result.

Each different age appears only once.

## 14. INNER JOIN

INNER JOIN combines records from two tables when they have matching values in a related column.

### Demonstration

For example:

A student can be connected to a course using `student_id`.

INNER JOIN will return only students who have a matching course.

```python
import sqlite3

connection = sqlite3.connect("college.db")
cursor = connection.cursor()

cursor.execute("""
SELECT students_pk.name, courses.course_name
FROM students_pk
INNER JOIN courses
ON students_pk.student_id = courses.student_id
""")

for row in cursor.fetchall():
    print(row)

connection.close()
```

### Explanation

The INNER JOIN connects the `students_pk` and `courses` tables using `student_id`.

Only records with matching student IDs are returned.

## 15. LEFT JOIN

LEFT JOIN returns all records from the left table and the matching records from the right table.

If there is no match, the right-side columns contain NULL.

### Demonstration

For example:

We can display all students, including students who are not enrolled in a course.

```python
import sqlite3

connection = sqlite3.connect("college.db")
cursor = connection.cursor()

cursor.execute("""
SELECT students_pk.name, courses.course_name
FROM students_pk
LEFT JOIN courses
ON students_pk.student_id = courses.student_id
""")

for row in cursor.fetchall():
    print(row)

connection.close()
```

### Explanation

The LEFT JOIN keeps every student from the `students_pk` table.

If a student does not have a matching course, the course name will be NULL.

## 16. RIGHT JOIN

RIGHT JOIN returns all records from the right table and matching records from the left table.

If there is no match, the left-side columns contain NULL.

### Demonstration

For example:

We can display all courses, including courses that do not have a matching student.

SQLite does not directly support RIGHT JOIN, so we can reverse the tables and use LEFT JOIN.

```python
import sqlite3

connection = sqlite3.connect("college.db")
cursor = connection.cursor()

cursor.execute("""
SELECT courses.course_name, students_pk.name
FROM courses
LEFT JOIN students_pk
ON courses.student_id = students_pk.student_id
""")

for row in cursor.fetchall():
    print(row)

connection.close()
```

### Explanation

The `courses` table is placed on the left and `students_pk` is joined to it.

This gives the practical result of a RIGHT JOIN because SQLite does not directly support RIGHT JOIN.

## 17. FULL JOIN

FULL JOIN returns all matching and non-matching records from both tables.

If a record has no match, the missing side contains NULL.

### Demonstration

For example:

A FULL JOIN can show:

- Students who have courses
- Students who do not have courses
- Courses that do not have students

```python
import sqlite3

connection = sqlite3.connect("college.db")
cursor = connection.cursor()

cursor.execute("""
SELECT students_pk.name, courses.course_name
FROM students_pk
LEFT JOIN courses
ON students_pk.student_id = courses.student_id

UNION

SELECT students_pk.name, courses.course_name
FROM courses
LEFT JOIN students_pk
ON students_pk.student_id = courses.student_id
""")

for row in cursor.fetchall():
    print(row)

connection.close()
```

### Explanation

SQLite does not directly support FULL JOIN.

Here, two LEFT JOIN queries are combined using UNION to include records from both tables, including unmatched records.

## 18. CROSS JOIN

CROSS JOIN creates every possible combination of rows from two tables.

### Demonstration

For example:

If there are 3 students and 2 courses:

3 × 2 = 6 combinations.

Each student will be paired with every course.

```python
import sqlite3

connection = sqlite3.connect("college.db")
cursor = connection.cursor()

cursor.execute("""
SELECT students_pk.name, courses.course_name
FROM students_pk
CROSS JOIN courses
""")

for row in cursor.fetchall():
    print(row)

connection.close()
```

### Explanation

The CROSS JOIN creates every possible student-course combination.

It does not require a matching condition between the tables.



## 19. UNION

UNION combines the results of two SELECT queries into one result.

It removes duplicate rows by default.

### Demonstration

For example:

We can combine names from two different tables into one result.

Both SELECT queries must have the same number of columns.

```python
import sqlite3

connection = sqlite3.connect("college.db")
cursor = connection.cursor()

cursor.execute("""
SELECT name FROM students_pk
UNION
SELECT course_name FROM courses
""")

for row in cursor.fetchall():
    print(row)

connection.close()
```

### Explanation

The UNION combines the results of the two SELECT statements into one result.

Duplicate values are removed automatically.

## 20. CASE

CASE is used to create conditional logic inside an SQL query.

It works similarly to if-else statements in Python.

### Demonstration

For example:

If a student is 18 or older, we can label them as "Adult".

Otherwise, we can label them as "Minor".

```python
import sqlite3

connection = sqlite3.connect("college.db")
cursor = connection.cursor()

cursor.execute("""
SELECT name, age,
CASE
    WHEN age >= 18 THEN 'Adult'
    ELSE 'Minor'
END AS age_group
FROM students
""")

for row in cursor.fetchall():
    print(row)

connection.close()
```

### Explanation

The CASE statement checks the student's age.

If the age is 18 or greater, it returns "Adult". Otherwise, it returns "Minor".

## 21. Subqueries

A subquery is a query written inside another SQL query.

The inner query produces a result that is used by the outer query.

### Demonstration

For example:

We can first find the average student age and then find students whose age is greater than that average.

```python
import sqlite3

connection = sqlite3.connect("college.db")
cursor = connection.cursor()

cursor.execute("""
SELECT name, age
FROM students
WHERE age > (
    SELECT AVG(age)
    FROM students
)
""")

for row in cursor.fetchall():
    print(row)

connection.close()
```

### Explanation

The inner query calculates the average age.

The outer query then returns students whose age is greater than that average.

## 22. Common Table Expressions (CTEs)

A Common Table Expression, or CTE, is a temporary named result that can be used inside a SQL query.

CTEs make complex queries easier to read and organize.

### Demonstration

Instead of writing a subquery directly inside the main query, we can first create a temporary result using `WITH`.

The main query can then use that result.

```python
import sqlite3

connection = sqlite3.connect("college.db")
cursor = connection.cursor()

cursor.execute("""
WITH average_age AS (
    SELECT AVG(age) AS avg_age
    FROM students
)
SELECT name, age
FROM students, average_age
WHERE age > avg_age
""")

for row in cursor.fetchall():
    print(row)

connection.close()
```

### Explanation

The CTE named `average_age` calculates the average student age.

The main query uses this temporary result to find students older than the average.

## 23. Views

A view is a saved SQL query that behaves like a virtual table.

It can be used to simplify frequently used queries.

### Demonstration

For example:

We can create a view containing only student names and ages.

Later, we can query the view instead of writing the original query again.

```python
import sqlite3

connection = sqlite3.connect("college.db")
cursor = connection.cursor()

cursor.execute("DROP VIEW IF EXISTS student_details")

cursor.execute("""
CREATE VIEW student_details AS
SELECT name, age
FROM students
""")

cursor.execute("SELECT * FROM student_details")

for row in cursor.fetchall():
    print(row)

connection.close()
```

### Explanation

The `student_details` view stores the SELECT query.

We can query the view like a table without rewriting the original SELECT statement.

## 24. Indexes

An index helps a database find rows faster when searching or filtering data.

Indexes can improve query performance, especially for large tables.

### Demonstration

For example:

If we frequently search students by name, we can create an index on the `name` column.

This can make searches faster.

```python
import sqlite3

connection = sqlite3.connect("college.db")
cursor = connection.cursor()

cursor.execute("""
CREATE INDEX IF NOT EXISTS idx_student_name
ON students(name)
""")

connection.commit()
connection.close()

print("Index created successfully.")
```

### Explanation

The index `idx_student_name` is created on the `name` column.

The database can use this index to make searches based on student names more efficient.

## 25. Window Functions

Window functions perform calculations across related rows while keeping each individual row in the result.

Unlike GROUP BY, window functions do not combine multiple rows into one row.

### Demonstration

For example:

We can display every student's age while also calculating their position or ranking among other students.

Each student's original row is still displayed.

```python
import sqlite3

connection = sqlite3.connect("college.db")
cursor = connection.cursor()

cursor.execute("""
SELECT name, age,
ROW_NUMBER() OVER (ORDER BY age DESC) AS row_number
FROM students
""")

for row in cursor.fetchall():
    print(row)

connection.close()
```

### Explanation

The `ROW_NUMBER()` window function assigns a number to each student based on age.

The original student rows are still returned.

## 26. ROW_NUMBER()

ROW_NUMBER() assigns a unique sequential number to each row in the result.

### Demonstration

If students are sorted by age from highest to lowest, ROW_NUMBER() gives them positions such as:

1 → First student  
2 → Second student  
3 → Third student

```python
import sqlite3

connection = sqlite3.connect("college.db")
cursor = connection.cursor()

cursor.execute("""
SELECT name, age,
ROW_NUMBER() OVER (ORDER BY age DESC) AS position
FROM students
""")

for row in cursor.fetchall():
    print(row)

connection.close()
```

### Explanation

ROW_NUMBER() assigns a different number to every row based on the specified ordering.

## 27. RANK()

RANK() assigns a rank to each row based on a specified order.

Rows with the same value receive the same rank, and gaps can appear after tied ranks.

### Demonstration

If two students have the same age, they receive the same rank.

For example:

Age 25 → Rank 1  
Age 25 → Rank 1  
Age 24 → Rank 3

```python
import sqlite3

connection = sqlite3.connect("college.db")
cursor = connection.cursor()

cursor.execute("""
SELECT name, age,
RANK() OVER (ORDER BY age DESC) AS age_rank
FROM students
""")

for row in cursor.fetchall():
    print(row)

connection.close()
```

### Explanation

RANK() gives the same rank to students with the same age.

Because two students may share a rank, the next rank can contain a gap.

## 28. DENSE_RANK()

DENSE_RANK() ranks rows based on their values.

Rows with the same value receive the same rank, but unlike RANK(), there are no gaps.

### Demonstration

For example:

Age 25 → Rank 1  
Age 25 → Rank 1  
Age 24 → Rank 2

There is no skipped rank.

```python
import sqlite3

connection = sqlite3.connect("college.db")
cursor = connection.cursor()

cursor.execute("""
SELECT name, age,
DENSE_RANK() OVER (ORDER BY age DESC) AS dense_age_rank
FROM students
""")

for row in cursor.fetchall():
    print(row)

connection.close()
```

### Explanation

DENSE_RANK() gives the same rank to equal values.

Unlike RANK(), the next rank is not skipped.

## 29. LAG()

LAG() allows us to access the value from a previous row without using a self-join.

### Demonstration

For example:

When students are sorted by age, LAG() can show the age of the previous student.

This allows us to compare the current row with the previous row.

```python
import sqlite3

connection = sqlite3.connect("college.db")
cursor = connection.cursor()

cursor.execute("""
SELECT name, age,
LAG(age) OVER (ORDER BY age) AS previous_age
FROM students
""")

for row in cursor.fetchall():
    print(row)

connection.close()
```

### Explanation

LAG() retrieves the age from the previous row.

For the first row, there is no previous row, so the result is NULL.

## 30. LEAD()

LEAD() allows us to access the value from the next row.

### Demonstration

For example:

When students are sorted by age, LEAD() can show the age of the next student.

This allows us to compare the current row with the next row.

```python
import sqlite3

connection = sqlite3.connect("college.db")
cursor = connection.cursor()

cursor.execute("""
SELECT name, age,
LEAD(age) OVER (ORDER BY age) AS next_age
FROM students
""")

for row in cursor.fetchall():
    print(row)

connection.close()
```

### Explanation

LEAD() retrieves the age from the next row.

For the last row, there is no next row, so the result is NULL.
