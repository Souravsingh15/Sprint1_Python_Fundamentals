# Reading Text Files

Reading a text file means opening an existing text file and retrieving its contents using Python.

Python provides the `open()` function to open and read files.

## Reading a File

The `"r"` mode is used to open a file for reading.

The `read()` method reads the complete content of the file.

### Explanation

- `open()` → Opens the file.
- `"r"` → Opens the file in read mode.
- `read()` → Reads the complete file content.
- `with` → Automatically closes the file after use.

# Writing Text Files

Writing a text file means adding content to a file or creating a new file and storing data in it.

Python uses `"w"` mode to write data into a file.

## Writing to a File

The `"w"` mode opens a file for writing.

If the file does not exist, Python creates it.

If the file already exists, `"w"` mode replaces its existing content.

```python
with open("data.txt", "w") as file:
    file.write("Hello, this is my first file.")

print("Data written successfully.")
```

### Explanation

- `"w"` → Write mode.
- `write()` → Writes content to the file.
- If the file already exists, its previous content is replaced.

# CSV Files

CSV stands for Comma-Separated Values.

CSV files are commonly used to store tabular data such as names, ages, marks, and employee records.

Python provides the built-in `csv` module to work with CSV files.

## Reading a CSV File

The `csv.reader()` function reads rows from a CSV file.

import csv

with open("students.csv", "r") as file:
    reader = csv.reader(file)

    for row in reader:
        print(row)

### Explanation

- `csv` → Python module for working with CSV files.
- `csv.reader()` → Reads CSV rows.
- Each row is returned as a list.

# JSON Files

JSON stands for JavaScript Object Notation.

JSON is a lightweight format commonly used to store and exchange structured data.

Python provides the `json` module to work with JSON files.

## Writing JSON Data

Python dictionaries can be converted into JSON format using `json.dump()`.

```python
import json

student = {
    "name": "Sourav",
    "age": 26,
    "course": "Python"
}

with open("student.json", "w") as file:
    json.dump(student, file, indent=4)

print("JSON file created successfully.")
```

### Explanation

- `json.dump()` → Writes Python data into a JSON file.
- `indent=4` → Makes the JSON file easier to read.
- Python dictionaries are commonly stored as JSON objects.

# Pickle Files

Pickling is the process of converting a Python object into a byte stream so that it can be stored or saved.

Unpickling is the process of converting the byte stream back into the original Python object.

## Pickling and Unpickling

The `pickle.dump()` function is used to save an object.

The `pickle.load()` function is used to retrieve the object.

```python
import pickle

student = {
    "name": "Sourav",
    "age": 26
}

# Pickling
with open("student.pkl", "wb") as file:
    pickle.dump(student, file)

# Unpickling
with open("student.pkl", "rb") as file:
    data = pickle.load(file)

print(data)
```

# Context Manager (`with`)

A context manager manages resources automatically.

The `with` statement is commonly used for file handling because it automatically closes the file after the operation is completed.

## Using `with`

Using `with` is safer and cleaner than manually opening and closing a file.

```python
with open("data.txt", "r") as file:
    data = file.read()

print(data)
```

### Explanation

The `with` statement automatically closes the file after the block finishes, even if an error occurs.

# pathlib

`pathlib` is a Python module used to work with files and directories using an object-oriented approach.

It provides an easier and more readable way to handle file paths.

```python
from pathlib import Path

file_path = Path("data.txt")

print(file_path.exists())
print(file_path.name)
```

### Explanation

- `Path()` → Creates a path object.
- `exists()` → Checks whether the path exists.
- `name` → Returns the file name.

# os Module

The `os` module provides functions for interacting with the operating system.

It can be used to work with files, directories, paths, and environment information.

## Listing Files

`os.listdir()` returns the files and directories present in a specified location.

```python
import os

files = os.listdir(".")

print(files)
```

### Explanation

`os.listdir()` returns the contents of the current directory.

`.` represents the current directory.

# shutil Module

The `shutil` module provides high-level operations for working with files and directories.

It can be used to copy, move, and remove files or directories.

## Copying a File

The `shutil.copy()` function copies a file from one location to another.

```python
import shutil

shutil.copy("data.txt", "data_copy.txt")

print("File copied successfully.")
```

### Explanation

- `shutil.copy()` → Copies a file.
- The first argument is the source.
- The second argument is the destination.

# File Validation

File validation means checking whether a file exists and whether it meets certain requirements before performing an operation.

Python provides functions such as `exists()` and `is_file()` for validation.

## Validating a File

Before reading a file, we can check whether it exists and is actually a file.

```python
from pathlib import Path

file_path = Path("data.txt")

if file_path.exists() and file_path.is_file():
    print("File is valid.")
else:
    print("File does not exist.")
```

### Explanation

- `exists()` → Checks whether the path exists.
- `is_file()` → Checks whether the path represents a file.
- Validation helps prevent file-related errors.

# Directory Operations

Directory operations involve creating, listing, renaming, and removing directories.

Python provides modules such as `os`, `pathlib`, and `shutil` for directory operations.

## Creating a Directory

The `mkdir()` method can be used to create a new directory.

```python
from pathlib import Path

folder = Path("my_folder")

folder.mkdir(exist_ok=True)

print("Directory created successfully.")
```

### Explanation

- `Path()` → Creates a path object.
- `mkdir()` → Creates a directory.
- `exist_ok=True` → Prevents an error if the directory already exists.
